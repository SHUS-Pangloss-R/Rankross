```python
import numpy as np
def generate_points(sigma,N=2000,R=5):
    n0=np.array([1,-2,2])/3.0
    a=np.array([1.0,0.0,0.0])
    u=np.cross(n0,a)
    u=u/np.linalg.norm(u)
    v=np.cross(n0,u)
    v=v/np.linalg.norm(v)
    r=np.sqrt(np.random.uniform(0,R**2,N))
    theta=np.random.uniform(0,2*np.pi,N)
    x=r*np.cos(theta)
    y=r*np.sin(theta)
    points=x[:,np.newaxis]*u+y[:,np.newaxis]*v
    noise=np.random.normal(0,sigma,(N,1))*n0
    points=points+noise
    return points,n0
def estimate_normal(points,n0):
    centroid=np.mean(points,axis=0)
    points_centered=points-centroid
    N=points.shape[0]
    c=(points_centered.T@points_centered)/N
    w,v=np.linalg.eigh(c)
    w=w[::-1]
    v=v[:,::-1]
    special=v[:,2]
    cos=np.abs(np.dot(special,n0))
    angle=np.degrees(np.arccos(np.clip(cos,-1.0,1.0)))
    planarity=(w[1]-w[2])/w[0]
    return w,angle,planarity
def run(sigma):
    print(f"noise={sigma}")
    points,n0=generate_points(sigma)
    w,angle,planarity=estimate_normal(points,n0)
    print(f"{w}")
    print(f"{angle}")
    print(f"{planarity}")
if __name__ == "__main__":
    np.random.seed(42)
    run(sigma=0.05)
    run(sigma=0.5)
    
```

```python
import open3d as o3d
import numpy as np
ply_point_cloud = o3d.data.PLYPointCloud()
pcd = o3d.io.read_point_cloud(ply_point_cloud.path)
print(f"原始点云点数: {len(pcd.points)}")

cl, ind = pcd.remove_statistical_outlier(nb_neighbors=20, std_ratio=2.0)
pcd_clean = pcd.select_by_index(ind)
print(f"去噪后点数: {len(pcd_clean.points)}")

downpcd = pcd_clean.voxel_down_sample(voxel_size=0.001)
print(f"下采样后点数: {len(downpcd.points)}")

z_coords = np.asarray(downpcd.points)[:, 2]

z_normalized = (z_coords - z_coords.min()) / (z_coords.max() - z_coords.min())

import matplotlib.pyplot as plt
colors = plt.cm.plasma(z_normalized)[:, :3] 
downpcd.colors = o3d.utility.Vector3dVector(colors)

print("正在打开可视化窗口...")
vis = o3d.visualization.Visualizer()
vis.create_window(window_name="Beautiful Point Cloud", width=1024, height=768)
vis.add_geometry(downpcd)
render_option = vis.get_render_option()
render_option.point_size = 2.0 
render_option.background_color = np.array([0.1, 0.1, 0.1]) 
vis.run()
vis.destroy_window()
# A little bit beautiful. a pity that all the normals were gone
```
